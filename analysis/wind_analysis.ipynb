```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#create dataset
data = {
    "actual": [12223,12804,12884,13493,14065,14432,14777,15468,16263,16775],
    "forecast": [19083,19250,19900,20245,20587,20396,20763,20802,19034,19090]
}

df = pd.DataFrame(data)

df
```

```python
#compute forecasat error
df["error"] = df["forecast"] - df["actual"]

df
```

```python
#ErrorStatistics
print("Mean Error:", df["error"].mean())
print("Median Error:", df["error"].median())
print("P99 Error:", df["error"].quantile(0.99))
```

```python
#Actual vs forecast plot
plt.figure(figsize=(8,4))

plt.plot(df["actual"], label="Actual Wind")
plt.plot(df["forecast"], label="Forecast Wind")

plt.title("Actual vs Forecast Wind Generation")
plt.xlabel("Time Index")
plt.ylabel("MW")

plt.legend()
plt.show()
```

```python
#error over time
plt.figure(figsize=(8,4))

plt.plot(df["error"])

plt.title("Forecast Error Over Time")
plt.xlabel("Time Index")
plt.ylabel("MW Error")

plt.show()
```

```python
#error distribution
sns.histplot(df["error"], kde=True)

plt.title("Distribution of Forecast Errors")
plt.show()
```
